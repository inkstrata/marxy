# Tide heights at the harbour

Readings from the gauge for 12 October 2026, taken every half hour. The question is how far low water falls below chart datum on a spring tide.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("gauge-2026-10-12.csv", parse_dates=["at"])
df.head()
```

## Low water

The lowest reading of the day is the one to compare with chart datum.

```python
low = df.loc[df.height_m.idxmin()]
print(f"Low water {low.height_m:.2f} m at {low['at']:%H:%M}")
```

```python
plt.figure(figsize=(6, 2.6))
plt.plot(df["at"], df["height_m"], color="#2e6ebe")
plt.axhline(0, color="#555", linewidth=0.8, linestyle="--")
plt.title("Gauge height, 12 October 2026")
plt.show()
```

```python
from tidekit import embed
embed.chart(df)
```

```python
df.set_index("at").height_m.resample("1h").mean().loc["2026-10-13"]
```

Spring tides fall about 0.3 m below datum here. Next step: compare with the 26 October spring tide.
